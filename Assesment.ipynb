```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
np.set_printoptions(suppress=True)

base_dir = Path.cwd()
```

```python
csv_path = base_dir / "Pre-Launch Case Payment Report - EHR Report.csv"
if not csv_path.exists():
    csv_path = Path(r"h:\All My AI Interview Task\All My AI Interview Task\Augmedix\Pre-Launch Case Payment Report - EHR Report.csv")

df = pd.read_csv(csv_path)

print(df.head())
```

```python
df.columns
```

```python
unique_payers = df["Payer"].unique()
unique_payers
```

```python
payer_summary = df.groupby('Payer').agg(
    Total_Payer_Payments=('Payer Payment', 'sum'),
    Total_Billed_Charges=('Billed Charge', 'sum')
).reset_index()

payer_summary['Payment_to_Charge_%'] = (
    payer_summary['Total_Payer_Payments'] / payer_summary['Total_Billed_Charges'] * 100
)


payer_summary
```

✅ Deliverable 2: PAYER – NPI – TIN MAPPING

```python
unique_npi_tin = df[['NPI', 'TIN']].drop_duplicates()
unique_npi_tin
```

```python
mapping = unique_npi_tin.merge(df[['Payer']].drop_duplicates(), how='cross')
mapping
```

```python
output_path = base_dir / "Shakil_Ahamed_01771604501.xlsx"

with pd.ExcelWriter(output_path, engine='xlsxwriter') as writer:
    payer_summary.to_excel(writer, sheet_name="Payer Summary", index=False)
    mapping.to_excel(writer, sheet_name="Payer-NPI-TIN Mapping", index=False)

output_path
```
